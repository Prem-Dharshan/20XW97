# OpenCV Basics

A guided walkthrough of the fundamentals used throughout this repo — image I/O, properties, transforms, color spaces, drawing, and direct pixel access. Every image here is just a NumPy array.

```python
import cv2 as cv
import numpy as np
```

## 1. Creating & Saving an Image

An image is a `(height, width, channels)` array of `uint8`. `cv.imwrite` picks the codec from the file extension.

```python
img = np.zeros((120, 400, 3), dtype=np.uint8)
cv.putText(img, "Hello, OpenCV!", (10, 80), cv.FONT_HERSHEY_SIMPLEX, 2, (255, 255, 255), 3)

cv.imshow("OpenCV OK", img)
cv.waitKey(0)
cv.destroyAllWindows()

cv.imwrite("./hello.png", img)
```

## 2. Reading Images

`cv.imread` loads color (BGR, not RGB) by default, or grayscale with `cv.IMREAD_GRAYSCALE` / flag `0`.

```python
img = cv.imread("../images/roi.jpg")
img_gray = cv.imread("../images/roi.jpg", cv.IMREAD_GRAYSCALE)

cv.imshow("Color", img)
cv.imshow("Grayscale", img_gray)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 3. Image Properties

`.shape` is `(height, width, channels)` — grayscale images drop the channel axis.

```python
print("Color shape:", img.shape)
print("Grayscale shape:", img_gray.shape)
print("Dtype:", img.dtype)
```

## 4. Resizing

Pass a target size, or leave it `(0, 0)` and scale with `fx`/`fy`.

```python
resized = cv.resize(img, (400, 300))
shrunk = cv.resize(img, (0, 0), fx=0.5, fy=0.5)
scaled = cv.resize(img, (0, 0), fx=2, fy=2)

cv.imshow("Resized", resized)
cv.imshow("Shrunk", shrunk)
cv.imshow("Scaled", scaled)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 5. Cropping (ROI)

Images slice like any NumPy array: `img[y1:y2, x1:x2]`.

```python
crop = img[50:200, 100:300]

cv.imshow("Original", img)
cv.imshow("Crop", crop)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 6. Rotating & Flipping

```python
rotated = cv.rotate(img, cv.ROTATE_90_COUNTERCLOCKWISE)
flipped = cv.flip(img, 1)

cv.imshow("Rotated", rotated)
cv.imshow("Flipped", flipped)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 7. Color Spaces

OpenCV stores color as BGR. Convert with `cv.cvtColor`.

```python
gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)
hsv = cv.cvtColor(img, cv.COLOR_BGR2HSV)

cv.imshow("Gray", gray)
cv.imshow("HSV", hsv)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 8. Splitting & Merging Channels

```python
b, g, r = cv.split(img)
merged = cv.merge([b, g, r])

cv.imshow("Blue Channel", b)
cv.imshow("Green Channel", g)
cv.imshow("Red Channel", r)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 9. Drawing Shapes & Text

```python
canvas = np.zeros((300, 300, 3), dtype=np.uint8)

cv.rectangle(canvas, (30, 30), (150, 150), (0, 255, 0), 2)
cv.circle(canvas, (220, 90), 60, (255, 0, 0), -1)
cv.line(canvas, (0, 200), (300, 200), (0, 0, 255), 3)
cv.putText(canvas, "Canvas", (80, 280), cv.FONT_HERSHEY_SIMPLEX, 1, (255, 255, 255), 2)

cv.imshow("Canvas", canvas)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 10. Arithmetic & Blending

`cv.add`/`cv.subtract` saturate at 0/255 — plain NumPy `+`/`-` on `uint8` wraps around instead. `cv.addWeighted` blends two images.

```python
img2 = cv.imread("../images/lenna.png")
img2 = cv.resize(img2, (img.shape[1], img.shape[0]))

added = cv.add(img, img2)
subtracted = cv.subtract(img, img2)
blended = cv.addWeighted(img, 0.7, img2, 0.3, 0)

cv.imshow("Added", added)
cv.imshow("Subtracted", subtracted)
cv.imshow("Blended", blended)
cv.waitKey(0)
cv.destroyAllWindows()
```

## 11. Pixel Access & Direct NumPy Indexing

Since images are arrays, pixels can be read and written directly — the same technique the manual, from-scratch algorithms in `concepts/` build on.

```python
arr = np.array(
    [
        [[0, 0, 0], [255, 255, 255]],
        [[255, 255, 255], [0, 0, 0]]
    ],
    dtype=np.uint8)

big = cv.resize(arr, (200, 200), interpolation=cv.INTER_NEAREST)

print("Pixel (0,0):", big[0, 0])
print("Row 199, cols 45:50:", big[199, 45:50])

big[0:100, 0:100] = [255, 0, 0]

cv.imshow("Array Image", big)
cv.waitKey(0)
cv.destroyAllWindows()
```
