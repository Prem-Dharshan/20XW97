```python
import cv2 as cv
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
lenna_path = '../images/lenna.png'
img = cv.imread(lenna_path)

cv.imshow("Lenna", img)
cv.waitKey(0)
cv.destroyAllWindows()
```

Write a program to resize an image using cv2, print the sizes of the original image size and resized image.

```python
path1 = "../images/lenna/half_lenna.png"
path2 = "../images/lenna/double_lenna.png"

half_resized_img = cv.resize(img, dsize=(0, 0), fx=0.5, fy=0.5)
double_resized_img = cv.resize(img, dsize=(0, 0), fx=2, fy=2)

cv.imwrite(path1, half_resized_img)
cv.imwrite(path2, double_resized_img)

cv.imshow(f"Original = {os.path.getsize('../images/lenna.png')} bytes", img)
cv.imshow(f"50% = {os.path.getsize(path1)} bytes", half_resized_img)
cv.imshow(f"200% = {os.path.getsize(path2)} bytes", double_resized_img)

cv.waitKey(0)
cv.destroyAllWindows()
```

Write a program to crop an image.
The corner points can be specified directly.
Use mouse to take points on the image.

```python
y1 = 100
y2 = 200
x1 = 50
x2 = 200

crop = img[y1:y2, x1:x2]

cv.imshow("Original", img)
cv.imshow("Crop", crop)

cv.waitKey(0)
cv.destroyAllWindows()
```

Write a program to  translate an image using `cv2.warpAffine` and using `imutils`

```python
import imutils

height, width = img.shape[:2]

matrix = np.array([
    [1, 0, 50],
    [0, 1, 50]
], dtype=np.float32)

translated = cv.warpAffine(img, matrix, (width, height))

cv.imshow("Original", img)
cv.imshow("warpAffine", translated)

translated = imutils.translate(img, 50, 30)

cv.imshow("imutils", translated)

cv.waitKey(0)
cv.destroyAllWindows()
```

Write a program to rotate  image
use fixed 900 rotation using cv2 constants
any arbitrary angle using cv2.getRotationMatrix2D()


```python
# Fixed rotations
rot_90 = cv.rotate(img, cv.ROTATE_90_CLOCKWISE)
rot_270 = cv.rotate(img, cv.ROTATE_90_COUNTERCLOCKWISE)
rot_180 = cv.rotate(img, cv.ROTATE_180)

# Arbitrary rotations
height, width = img.shape[:2]
center = (width // 2, height // 2)

matrix_45 = cv.getRotationMatrix2D(center, 45, 1)
matrix_135 = cv.getRotationMatrix2D(center, 135, 1)

rot_45 = cv.warpAffine(img, matrix_45, (width, height))
rot_135 = cv.warpAffine(img, matrix_135, (width, height))

cv.imshow("Original", img)
cv.imshow("90 Clockwise", rot_90)
cv.imshow("90 Counterclockwise", rot_270)
cv.imshow("180", rot_180)
cv.imshow("45", rot_45)
cv.imshow("135", rot_135)

cv.waitKey(0)
cv.destroyAllWindows()
```

Write a program to flip an  image in x-direction, y-direction and xy-direction.

```python
x_flip = cv.flip(img, 0)
y_flip = cv.flip(img, 1)
xy_flip = cv.flip(img, -1)

cv.imshow("Original", img)
cv.imshow("X Direction", x_flip)
cv.imshow("Y Direction", y_flip)
cv.imshow("XY Direction", xy_flip)

cv.waitKey(0)
cv.destroyAllWindows()
```
