```python
import cv2 as cv
import numpy as np
import os
import matplotlib.pyplot as plt
```

```python
def show(images):
    plt.figure(figsize=(4 * len(images), 4))
    for i, (title, im) in enumerate(images.items()):
        plt.subplot(1, len(images), i + 1)
        if im.ndim == 2:
            plt.imshow(im, cmap="gray")
        else:
            plt.imshow(cv.cvtColor(im, cv.COLOR_BGR2RGB))
        plt.title(title)
        plt.xticks([]), plt.yticks([])
    plt.tight_layout()
    plt.show()
```

```python
lenna_path = '../images/lenna.png'
img = cv.imread(lenna_path)

show({"Lenna": img})
```

Write a program to resize an image using cv2, print the sizes of the original image size and resized image.

```python
path1 = "../images/lenna/half_lenna.png"
path2 = "../images/lenna/double_lenna.png"

half_resized_img = cv.resize(img, dsize=(0, 0), fx=0.5, fy=0.5)  # scale < 1 shrinks the image, discarding detail
double_resized_img = cv.resize(img, dsize=(0, 0), fx=2, fy=2)  # scale > 1 enlarges the image, interpolating new pixels (no new detail added)

cv.imwrite(path1, half_resized_img)
cv.imwrite(path2, double_resized_img)

show({
    f"Original = {os.path.getsize('../images/lenna.png')} bytes": img,
    f"50% = {os.path.getsize(path1)} bytes": half_resized_img,
    f"200% = {os.path.getsize(path2)} bytes": double_resized_img,
})
```

Write a program to crop an image.
The corner points can be specified directly.
Use mouse to take points on the image.

```python
y1 = 100
y2 = 200
x1 = 50
x2 = 200  # the (x1, x2, y1, y2) box selects which region is kept; moving/resizing the box changes what part of the image survives the crop

crop = img[y1:y2, x1:x2]

show({"Original": img, "Crop": crop})
```

Write a program to  translate an image using `cv2.warpAffine` and using `imutils`

```python
import imutils

height, width = img.shape[:2]

matrix = np.array([
    [1, 0, 50],  # tx = 50 -> shifts the image 50px right (larger tx shifts further right)
    [0, 1, 50]   # ty = 50 -> shifts the image 50px down (larger ty shifts further down)
], dtype=np.float32)

translated_warp = cv.warpAffine(img, matrix, (width, height))
translated_imutils = imutils.translate(img, 50, 30)  # same idea: (x, y) offsets shift the image right/down by that many pixels

show({"Original": img, "warpAffine": translated_warp, "imutils": translated_imutils})
```

Write a program to rotate  image
use fixed 900 rotation using cv2 constants
any arbitrary angle using cv2.getRotationMatrix2D()


```python
# Fixed rotations
rot_90 = cv.rotate(img, cv.ROTATE_90_CLOCKWISE)
rot_270 = cv.rotate(img, cv.ROTATE_90_COUNTERCLOCKWISE)
rot_180 = cv.rotate(img, cv.ROTATE_180)

# Arbitrary rotations
height, width = img.shape[:2]
center = (width // 2, height // 2)

matrix_45 = cv.getRotationMatrix2D(center, 45, 1)  # rotation angle in degrees: larger angle rotates the image further counter-clockwise
matrix_135 = cv.getRotationMatrix2D(center, 135, 1)

print(matrix_135)

rot_45 = cv.warpAffine(img, matrix_45, (width, height))
rot_135 = cv.warpAffine(img, matrix_135, (width, height))

show({
    "Original": img,
    "90 Clockwise": rot_90,
    "90 Counterclockwise": rot_270,
    "180": rot_180,
    "45": rot_45,
    "135": rot_135,
})
```

Write a program to flip an  image in x-direction, y-direction and xy-direction.

```python
x_flip = cv.flip(img, 0)   # flip code 0 -> flip around the x-axis (vertical flip, top/bottom mirrored)
y_flip = cv.flip(img, 1)   # flip code 1 -> flip around the y-axis (horizontal flip, left/right mirrored)
xy_flip = cv.flip(img, -1)  # flip code -1 -> flip around both axes (mirrored both ways)

show({"Original": img, "X Direction": x_flip, "Y Direction": y_flip, "XY Direction": xy_flip})
```
